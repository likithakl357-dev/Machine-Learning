```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier

data = {
    "Weather":["Sunny","Sunny","Overcast","Rain","Rain","Rain","Overcast",
               "Sunny","Sunny","Rain","Sunny","Overcast","Overcast","Rain"],

    "Temperature":["Hot","Hot","Hot","Mild","Cool","Cool","Cool",
                   "Mild","Cool","Mild","Mild","Mild","Hot","Mild"],

    "Humidity":["High","High","High","High","Normal","Normal","Normal",
                "High","Normal","Normal","Normal","High","Normal","High"],

    "Class":["No","No","Yes","Yes","Yes","No","Yes",
             "No","Yes","Yes","Yes","Yes","Yes","No"]
}

df = pd.DataFrame(data)

encoder = LabelEncoder()

df["Weather"] = encoder.fit_transform(df["Weather"])
df["Temperature"] = encoder.fit_transform(df["Temperature"])
df["Humidity"] = encoder.fit_transform(df["Humidity"])
df["Class"] = encoder.fit_transform(df["Class"])

X = df[["Weather","Temperature","Humidity"]]
y = df["Class"]

model = DecisionTreeClassifier()

model.fit(X,y)

prediction = model.predict([[2,0,0]])

if prediction[0] == 1:
    print("Prediction : Play")
else:
    print("Prediction : Don't Play")
```

# Decision Tree with Train-Test **Split**

```python
# Step 1: Import libraries

import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Step 2: Create dataset

data = {
    "Weather":["Sunny","Sunny","Overcast","Rain","Rain","Rain","Overcast",
               "Sunny","Sunny","Rain","Sunny","Overcast","Overcast","Rain"],

    "Temperature":["Hot","Hot","Hot","Mild","Cool","Cool","Cool",
                   "Mild","Cool","Mild","Mild","Mild","Hot","Mild"],

    "Humidity":["High","High","High","High","Normal","Normal","Normal",
                "High","Normal","Normal","Normal","High","Normal","High"],

    "Class":["No","No","Yes","Yes","Yes","No","Yes",
             "No","Yes","Yes","Yes","Yes","Yes","No"]
}

df = pd.DataFrame(data)

# Step 3: Convert text into numbers

encoder = LabelEncoder()

df["Weather"] = encoder.fit_transform(df["Weather"])
df["Temperature"] = encoder.fit_transform(df["Temperature"])
df["Humidity"] = encoder.fit_transform(df["Humidity"])
df["Class"] = encoder.fit_transform(df["Class"])

# Step 4: Select input and output

X = df[["Weather","Temperature","Humidity"]]
y = df["Class"]

# Step 5: Split into Training and Testing data

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Step 6: Create Decision Tree Model

model = DecisionTreeClassifier()

# Step 7: Train the model

model.fit(X_train, y_train)

# Step 8: Predict the test data

y_pred = model.predict(X_test)

# Step 9: Display Actual and Predicted values

print("Actual Values")
print(y_test.values)

print("\nPredicted Values")
print(y_pred)

# Step 10: Calculate Accuracy

accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy =", round(accuracy*100,2), "%")
```
