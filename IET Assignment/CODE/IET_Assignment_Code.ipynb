```python
df = pd.read_csv("AshokVihar_Hourly.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
if "Unnamed: 0" in df.columns:
    df = df.drop("Unnamed: 0", axis=1)

print("Updated dataset shape:", df.shape)
print("\nUpdated columns:")
print(df.columns.tolist())
```

```python
df.describe()
```

```python
print("PM2.5 Statistics")
print("------------------")

print("Mean:", df["PM2.5"].mean())
print("Median:", df["PM2.5"].median())
print("Standard Deviation:", df["PM2.5"].std())
print("Minimum:", df["PM2.5"].min())
print("Maximum:", df["PM2.5"].max())
```

```python
pollutants = [
    "PM2.5",
    "PM10",
    "NO",
    "NO2",
    "SO2",
    "Ozone",
    "CO",
    "Benzene",
    "NH3",
    "NOx"
]

for column in pollutants:
    negative_values = (df[column] < 0).sum()
    print(f"{column}: {negative_values} negative values")
```

```python
df["datetime"] = pd.to_datetime(
    df[["year", "month", "day", "hour"]]
)

print(df[["datetime"]].head())
```

```python
df = df.sort_values("datetime").reset_index(drop=True)

print("Dataset sorted chronologically.")
print(df[["datetime", "PM2.5"]].head())
```

```python
print("FINAL DATASET CHECK")
print("===================")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing values:", df.isnull().sum().sum())

print("Duplicate rows:", df.duplicated().sum())

print("\nColumns:")
print(df.columns.tolist())
```

```python
plt.figure(figsize=(15, 5))

plt.plot(
    df["datetime"],
    df["PM2.5"]
)

plt.title("PM2.5 Concentration Over Time")
plt.xlabel("Date")
plt.ylabel("PM2.5 Concentration")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["PM2.5"],
    bins=40
)

plt.title("Distribution of PM2.5 Concentration")
plt.xlabel("PM2.5 Concentration")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
major_pollutants = [
    "PM2.5",
    "PM10",
    "NO2",
    "SO2",
    "Ozone",
    "CO"
]

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df[major_pollutants]
)

plt.title("Boxplot of Major Air Pollutants")
plt.xlabel("Pollutant")
plt.ylabel("Concentration")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 10))

correlation = df.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix of Air Quality Variables")

plt.tight_layout()
plt.show()
```

```python
hourly_pm25 = df.groupby("hour")["PM2.5"].mean()

plt.figure(figsize=(10, 5))

plt.plot(
    hourly_pm25.index,
    hourly_pm25.values,
    marker="o"
)

plt.title("Average PM2.5 Concentration by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average PM2.5")

plt.xticks(range(24))

plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
monthly_pm25 = df.groupby("month")["PM2.5"].mean()

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_pm25.index,
    monthly_pm25.values,
    marker="o"
)

plt.title("Average PM2.5 Concentration by Month")
plt.xlabel("Month")
plt.ylabel("Average PM2.5")

plt.xticks(range(1, 13))

plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
df.to_csv(
    "AshokVihar_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
