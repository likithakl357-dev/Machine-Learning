```python
import pandas as pd
import numpy as np

# Load dataset
data = pd.read_csv('data.csv')
print(data)

# Separate features and target
concepts = np.array(data.iloc[:, :-1])
target = np.array(data.iloc[:, -1])

# Initialize most specific hypothesis
h = ['Ø'] * len(concepts[0])

print("\nInitial Hypothesis:", h)

# FIND-S Algorithm
for i, row in enumerate(concepts):
    if target[i] == "Yes":  # Only positive examples
        for x in range(len(h)):
            if h[x] == 'Ø':
                h[x] = row[x]
            elif h[x] != row[x]:
                h[x] = '?'

    print(f"\nStep {i+1}")
    print("Hypothesis:", h)

print("\nFinal Hypothesis:", h)
```
