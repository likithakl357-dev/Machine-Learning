```python
csv_data = """Outlook,Temperature,Humidity,Wind,PlayTennis
Sunny,Hot,High,Weak,No
Sunny,Hot,High,Strong,No
Overcast,Hot,High,Weak,Yes
Rain,Mild,High,Weak,Yes
Rain,Cool,Normal,Weak,Yes
Rain,Cool,Normal,Strong,No
Overcast,Mild,Normal,Strong,Yes
Sunny,Cool,High,Weak,No"""

with open('training_data.csv', 'w') as f:
    f.write(csv_data)
```

```python
import csv

# Load CSV file
data = []
with open('training_data.csv') as file:
    reader = csv.reader(file)
    next(reader)  # skip header
    for row in reader:
        data.append(row)

print("Training Data:\n", data)

# Initialize S and G
num_attributes = len(data[0]) - 1

S = ['0'] * num_attributes
G = [['?'] * num_attributes]

print("\nInitial S:", S)
print("Initial G:", G)

# Candidate-Elimination Algorithm
for i, row in enumerate(data):
    if row[-1] == "Yes":  # Positive example
        for j in range(num_attributes):
            if S[j] == '0':
                S[j] = row[j]
            elif S[j] != row[j]:
                S[j] = '?'

        # Remove inconsistent hypotheses from G
        G = [g for g in G if all(g[j] == '?' or g[j] == S[j] for j in range(num_attributes))]

    else:  # Negative example
        new_G = []
        for g in G:
            for j in range(num_attributes):
                if g[j] == '?':
                    if S[j] != row[j]:
                        new_h = g.copy()
                        new_h[j] = S[j]
                        new_G.append(new_h)

        G = new_G

    print(f"\nStep {i+1}")
    print("S:", S)
    print("G:", G)

# Final Output
print("\nFinal Specific Hypothesis (S):", S)
print("Final General Hypothesis (G):", G)
```
